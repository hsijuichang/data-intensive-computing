# Week 4 Task 4 — Spark Start-up Baseline

This notebook only starts a Spark session with the same configuration as `w4_task1.ipynb` and `w4_task4_approach_a_raw_dataset.ipynb`, then runs one trivial action.

The Task 4 timing harness runs every notebook in a fresh kernel, so each measurement includes kernel start-up, JVM start-up, Delta package resolution, and Spark session creation. Timing this notebook measures that fixed cost on its own, which separates it from the data-processing time of each approach.

```python
from pathlib import Path
import sys


def find_project_root(start: Path) -> Path:
    for candidate in (start, *start.parents):
        if (candidate / "pyproject.toml").is_file():
            return candidate
    raise FileNotFoundError("Could not find the project root containing pyproject.toml")


PROJECT_ROOT = find_project_root(Path.cwd().resolve())
source_package = str(PROJECT_ROOT / "src")
if source_package not in sys.path:
    sys.path.insert(0, source_package)

from data_intensive_computing.spark import build_spark_session

spark = build_spark_session(
    "Week4-Task4-Spark-Baseline",
    driver_memory="8g",
    shuffle_partitions=32,
)
print("Spark", spark.version, "on", spark.sparkContext.master)
print("Trivial action:", spark.range(1).count())
```
